# Observatory · Fast System

Abre una publicación de Athena sin entrenar, sin cargar pesos y sin necesitar el directorio del run. Conserva las vistas Historia y budgets, Model Arena y Alertamientos. Usa la configuración privada de publicación y el entorno del kernel actual.

```python
from pathlib import Path
import subprocess, sys
from importlib.metadata import version, PackageNotFoundError

VERSION = "0.1.4"
# La instalación conserva los índices y credenciales que ADA ya tenga configurados.
# Si el índice corporativo aún no replica la versión, usa el wheel adjunto.
WHEEL = next(iter(Path.cwd().glob("dist/model_arena_sdk-0.1.4-py3-none-any.whl")), None)
try:
    installed = version("model-arena-sdk")
except PackageNotFoundError:
    installed = None
if installed != VERSION:
    requirement = str(WHEEL.resolve()) if WHEEL else f"model-arena-sdk=={VERSION}"
    command = [sys.executable, "-m", "pip", "install", requirement]
    if WHEEL:
        command.extend(["--find-links", str(WHEEL.parent.resolve())])
    subprocess.run(command, check=True)
from model_arena import __version__
assert __version__ == VERSION, "Reinicia el kernel: quedó importada una versión anterior"
print("Model Arena", __version__, "· Python", sys.version.split()[0])
```

```python
from model_arena.profiles.mac import open_monitor

PUBLICATION_ID = None  # None permite elegir una publicación desde el monitor.
monitor = open_monitor(
    Path.cwd() / "private/publication.json",
    publication_id=PUBLICATION_ID,
    workspace=Path.cwd() / ".observatory",
)
print("PID:", monitor.pid)
```
